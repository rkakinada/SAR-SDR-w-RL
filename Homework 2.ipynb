```python
# Problem 7
w=2
b=0
x=[20,10,0]
y=[1,0,-1]
y_p = [0,0,0]
for i in range(len(x)-1):
    y_p[i] = np.tanh(2*x[i] + b)
print(y_p)
```

```python
# Problem 8
error=0
for i in range(len(y)):
    error = (y_p[i]-y[i])**2 + error 
print(error)
```

```python
def model(x_in,w_in,b_in):
    return torch.tanh(w_in*x_in+b_in)
def loss_func(y_p,y_in):
    return (y_p-y_in)**2
    
```

```python
# Problem 10
theta = torch.tensor([2.0, 0.0], requires_grad=True)
a=0.1
optim = torch.optim.SGD([theta],lr=a)
for i in range(len(x)):
    error = loss_func(model(x[i],*theta),y[i])
    optim.zero_grad()
    error.backward()
    optim.step()
    print(theta.grad)
# Problem 11
print(theta)
```
